```python
import random
import itertools
```

```python
unit = "m"
prepositions = [ ("k",3), ("",0), ("d",-1), ("c",-2), ("m",-3)]
```

```python
def cust_rnd_fmt(num):
    num = round(num,6)
    if num == int(num):
        n = f"{num:,.0f}"
    else:
        n = f"{num:,}"
    n = n.replace(","," ") .replace(".",",")   
    return n
```

```python
def get_pvals(preps, kms, ms, dms, cms, mms):
    pvals = []
    for p in preps:
        if p[0] == "k":
            pval = [p[0], p[1], kms]
            pvals.append(pval)
        if p[0] == "":
            pval = [p[0], p[1], ms]
            pvals.append(pval)        
        if p[0] == "d":
            pval = [p[0], p[1], dms]
            pvals.append(pval)  
        if p[0] == "c":
            pval = [p[0], p[1], cms]
            pvals.append(pval)    
        if p[0] == "m":
            pval = [p[0], p[1], mms]
            pvals.append(pval)  

    return (pvals)
```

```python
# m
def get_pvals_m():
    preps = random.sample([("k",3), ("d",-1), ("c",-2), ("m",-3)], 3)

    kms = random.randint(1,99)
    ms = 0
    dms = random.randint(1,999)
    cms = random.randint(1,999)
    mms = random.randint(1,999)

    if any(x[0] for x in preps if x[0] == "k"):
        if kms > 9:
            mms = round(mms,-2)
            cms = round(cms,-1)
        else:
            mms = round(mms,-1)  


    pvals = get_pvals(preps, kms, ms, dms, cms, mms)
    return (pvals)
```

```python

def get_pvals_km():
    preps = random.sample([("",0), ("d",-1), ("c",-2)], 3)

    kms = 0
    ms = random.randint(99,99999)
    dms = random.randint(1,99) * 10
    cms = random.randint(1,9) * 100
    mms = 0

    pvals = get_pvals(preps, kms, ms, dms, cms, mms)
    return (pvals)

    
```

```python
def get_pvals_dm():
    preps = random.sample([("k",3), ("",0), ("c",-2), ("m",-3) ], 3)

    kms = random.randint(1,9)
    ms = random.randint(1,999)
    dms = 0
    cms = random.randint(1,999)
    mms = random.randint(1,99) * 10

    pvals = get_pvals(preps, kms, ms, dms, cms, mms)
    return (pvals)
```

```python
def get_pvals_cm():
    preps = random.sample([ ("",0), ("d",-1), ("m",-3) ], 3)

    kms = 0
    ms = random.randint(1,99)
    dms = random.randint(1,999)
    cms = 0
    mms = random.randint(1,999)

    pvals = get_pvals(preps, kms, ms, dms, cms, mms)
    return (pvals)
```

```python
def get_pvals_mm():
    preps = random.sample([ ("",0), ("d",-1), ("c",-2) ], 3)

    kms = 0
    ms = random.randint(1,99)
    dms = random.randint(1,999)
    cms = random.randint(1,999)
    mms = 0

    pvals = get_pvals(preps, kms, ms, dms, cms, mms)
    return (pvals)
```

```python
def get_assgn(result_prep, pvals):
    result = 0
    txt = ""
    for e,i in enumerate(pvals):
        result += i[2] * 10**(i[1] -  result_prep[1])
        txt += str(i[2]) + " " + i[0] + unit
        if e < 2: 
            txt += " + "

        
    txt += " = " 
    txt += (28 - len(txt))*" "
    xxt = txt + 12*" " + result_prep[0] + unit    

    txt += cust_rnd_fmt(result) 
    txt += (37 - len(txt)) * " " + result_prep[0] + unit

    return (txt, xxt)
    
```

```python
assgns = []
res_assgns = []

for k in range(10):

    result_prep = ("k", 3)
    pvals = get_pvals_km()
    txt, xxt = get_assgn(result_prep, pvals)
    assgns.append(xxt)
    res_assgns.append(txt)

    result_prep = ("", 0)
    pvals = get_pvals_m()
    txt, xxt = get_assgn(result_prep, pvals)
    assgns.append(xxt)
    res_assgns.append(txt)    

    result_prep = ("d", -1)
    pvals = get_pvals_dm()
    txt, xxt = get_assgn(result_prep, pvals)
    assgns.append(xxt)
    res_assgns.append(txt)     

    result_prep = ("c", -2)
    pvals = get_pvals_cm()
    txt, xxt = get_assgn(result_prep, pvals)
    assgns.append(xxt)
    res_assgns.append(txt)     

    result_prep = ("m", -3)
    pvals = get_pvals_mm()
    txt, xxt = get_assgn(result_prep, pvals)
    assgns.append(xxt)
    res_assgns.append(txt)         


for  i,a in enumerate(assgns, start=1):
    if i < 10:
        sp = ".  "
    else: 
        sp = ". "        
    print (str(i) + sp + a)   

for  i,a in enumerate(res_assgns, start=1):
    if i < 10:
        sp = ".  "
    else: 
        sp = ". "        
    print (str(i) + sp + a)
```
