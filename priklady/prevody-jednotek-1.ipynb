```python
import random
import itertools
```

```python
prepositions = [ ("k",3), ("",0), ("d",-1), ("c",-2), ("m",-3)]
# Create a cyclic iterator over the tuples
cyclic_prepositions = itertools.cycle(prepositions)
unit = "m"

```

```python
def count_decimal_places(x):
    if not isinstance(x, float):
        return 0
    s = str(x).rstrip('0')          # remove trailing zeros
    if '.' not in s:
        return 0
    return len(s) - s.index('.') - 1
```

```python
def cust_rnd_fmt(num):
    num = round(num,6)
    if num == int(num):
        n = f"{num:,.0f}"
    else:
        n = f"{num:,}"
    n = n.replace(","," ") .replace(".",",")   
    return n
```

```python
assgns = []
res_assgns = []
def create_assgns(expected_prep_result):    

    while True:
        j = random.choice([0,1])
        preps = random.sample(prepositions[j:j+4], 3)
        numbers = random.sample(range(2, 99), 3)
        result = 0
        txt = ""
        for i in range(3):
            result += numbers[i] * 10**preps[i][1]
            txt += str(numbers[i]) + " " + preps[i][0] + unit
            if i < 2: 
                txt += " + "


        result = round(result,7)
        txt += " = " 
        txt += (25 - len(txt))*" "
        xxt = txt
        if result < 20:
            r = cust_rnd_fmt(result * 1000)
            txt +=  r + " " + "m" + unit
            xxt +=  12*" " + "m" + unit
        elif result < 99: 
            r = cust_rnd_fmt(result * 100)
            txt +=  r + " " + "c" + unit    
            xxt +=  12*" " + "c" + unit    
        elif result < 999: 
            r = cust_rnd_fmt(result * 10)
            txt +=  r + " " + "d" + unit    
            xxt +=  12*" " + "d" + unit        
        elif result < 9999: 
            r = cust_rnd_fmt(result * 1)
            txt +=  r + " " + "" + unit    
            xxt +=  12*" " + "" + unit           
        else:    
            r = cust_rnd_fmt(result / 1000) 
            txt +=  r + " " + "k" + unit   
            xxt +=  12*" " + "k" + unit   

        prep_result = txt[-2:-1].replace(" ","")
        print (expected_prep_result, prep_result, txt)
        
        if (prep_result == expected_prep_result) and result < 60_000:
            print("added")
            assgns.append(xxt)  
            res_assgns.append(txt)  
            
            break 
```

```python
assgns = []
res_assgns = []
for k in range(40):
    expected_prep_result = next((cyclic_prepositions))[0]
    create_assgns(expected_prep_result)

for  i,a in enumerate(assgns, start=1):
    if i < 10:
        sp = ".  "
    else: 
        sp = ". "        
    print (str(i) + sp + a)   

for  i,a in enumerate(res_assgns, start=1):
    if i < 10:
        sp = ".  "
    else: 
        sp = ". "        
    print (str(i) + sp + a)       
```
