```python
#upscaling

import random
import itertools
```

```python
prepositions = [ ("k",3), ("",0), ("d",-1), ("c",-2), ("m",-3)]
# Create a cyclic iterator over the tuples
cyclic_prepositions = itertools.cycle(prepositions)
unit = "m"
```

```python
def get_rndval_dm_cm_mm():
    mms = random.randint(11,99)*10
    dms = random.randint(101,999)

    comp_w = 1000 - (mms + 100 * dms) % 1000 # complement to next whole 1000 mm
    if comp_w == 1000:
        comp_w = 0
    cms = int(comp_w / 10) + 100*random.randint(1,9)
    return (dms, cms, mms)
```

```python
def get_rndval_m_dm_cm():
    m_intervals = [(1,1999), (1,9999), (1,99999)]
    low, high = random.choice(m_intervals)
    ms = random.randint(low,high)
    cms= random.randint(1,99)*10

    if ms > 9999:
        comp_w = 10_000 - (100 * ms + cms) % 10000 # complement to next whole 100 m in cm
        if comp_w == 10_000:
            comp_w = 0
    else:    
        comp_w = 1000 - (100 * ms + cms) % 1000 # complement to next whole 10 m in cm
        if comp_w == 1000:
            comp_w = 0

    dms =  int(comp_w / 10)    
    return (ms, dms, cms)
```

```python
get_rndval_m_dm_cm()
```

```python
def cust_rnd_fmt(num):
    num = round(num,6)
    if num == int(num):
        n = f"{num:,.0f}"
    else:
        n = f"{num:,}"
    n = n.replace(","," ") .replace(".",",")   
    return n
```

```python
def get_m_assgn():
    preps_dm_cm_mm = [ ("d",-1), ("c",-2), ("m",-3)]
    rndval_dm_cm_mm = get_rndval_dm_cm_mm()

    idxs = [0,1,2]
    random.shuffle(idxs)

    result = 0
    txt = ""
    for e,i in enumerate(idxs):
        result += rndval_dm_cm_mm[i] * 10**preps_dm_cm_mm[i][1]
        txt += str(rndval_dm_cm_mm[i]) + " " + preps_dm_cm_mm[i][0] + unit
        if e < 2: 
            txt += " + "

        
    txt += " = " 
    txt += (28 - len(txt))*" "
    xxt = txt + 12*" " + "" + unit    

    txt += cust_rnd_fmt(result) 
    txt += (33 - len(txt)) * " " + " m"

    return(xxt, txt)


```

```python
def get_km_assgn():
    preps_m_dm_cm = [ ("",0), ("d",-1), ("c",-2)]
    rndval_m_dm_cm = get_rndval_m_dm_cm()

    idxs = [0,1,2]
    random.shuffle(idxs)

    result = 0
    txt = ""
    for e,i in enumerate(idxs):
        result += rndval_m_dm_cm[i] * 10**preps_m_dm_cm[i][1]
        txt += str(rndval_m_dm_cm[i]) + " " + preps_m_dm_cm[i][0] + unit
        if e < 2: 
            txt += " + "

        
    txt += " = " 
    txt += (28 - len(txt))*" "
    xxt = txt + 12*" " + "k" + unit    

    txt += cust_rnd_fmt(result/1000) 
    txt += (33 - len(txt)) * " " + "km"

    return(xxt, txt)        

```

```python
assgns = []
res_assgns = []

for k in range(20):
    a = get_m_assgn()
    assgns.append(a[0])
    res_assgns.append(a[1])

    a = get_km_assgn()
    assgns.append(a[0])
    res_assgns.append(a[1])    


for  i,a in enumerate(assgns, start=1):
    if i < 10:
        sp = ".  "
    else: 
        sp = ". "        
    print (str(i) + sp + a)   

for  i,a in enumerate(res_assgns, start=1):
    if i < 10:
        sp = ".  "
    else: 
        sp = ". "        
    print (str(i) + sp + a) 


```
